## Dataset 1 - Pakistani E-Commerce Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("pakistani_ecommerce_dataset.csv")

# Display first five records
print(df.head())

# Display dataset information
print(df.info())

# Check for missing values
print(df.isnull().sum())
```

##### Q1. Which is the most popular product category?

```python
category_count = df["product_category"].value_counts()

print(category_count)

print("Most Popular Category:")
print(category_count.idxmax())
```

Home Decor is the most popular category, with 30 records.

##### Q2. What is the average order value?

```python
average_order_value = df["order_value"].mean()

print("Average Order Value:", round(average_order_value, 2))
```

The average order value is 2832.15 dataset currency units.

##### Q3. Which city has the highest number of orders?

```python
city_orders = df["city"].value_counts()

print(city_orders)

print("City with Highest Orders:")
print(city_orders.idxmax())
```

The city with the highest number of orders was Karachi, with 24 orders.

##### Q4. What is the most used payment method?

```python
payment_count = df["payment_method"].value_counts()

print(payment_count)

print("Most Used Payment Method:")
print(payment_count.idxmax())
```

Cash on Delivery is the most frequently used payment method, appearing 29 times.

##### Q5. Draw a bar chart showing average order value by city.

```python
average_by_city = df.groupby("city")["order_value"].mean()

print(average_by_city)

average_by_city.plot(
    kind="bar",
    figsize=(10, 5),
    color="skyblue"
)

plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value (PKR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Dataset 2 - Student Lifestyle & Study Habits

```python
students = pd.read_csv("student_lifestyle_dataset.csv")

print(students.head())

print(students.info())

print(students.isnull().sum())
```

##### Q1. What is the average number of study hours?

```python
average_study = students["study_hours"].mean()

print(f"Average Study Hours: {average_study:.2f}")
```

The average number of study hours is approximately 4.40 hours.

##### Q2. Who sleeps more on average - males or females?

```python
sleep_by_gender = students.groupby("gender")["sleep_hours"].mean()

print(sleep_by_gender)

print("Students who sleep more on average:")
print(sleep_by_gender.idxmax())
```

Female students sleep more on average.

##### Q3. Is there a relationship between screen time and sleep hours?

```python
correlation = students["screen_time"].corr(
    students["sleep_hours"]
)
print("Correlation:", round(correlation, 3))
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=students,
    x="screen_time",
    y="sleep_hours"
)

plt.title("Relationship Between Screen Time and Sleep Hours")
plt.xlabel("Screen Time")
plt.ylabel("Sleep Hours")

plt.show()
```

The correlation is approximately 0.052. This indicates a negligible positive linear relationship between screen time and sleep hours. In other words, the dataset does not show a meaningful linear association between these two variables.

##### Q4. Draw a histogram of daily transport cost.

```python
plt.figure(figsize=(8, 5))

plt.hist(
    students["daily_transport_cost"],
    bins=10,
    edgecolor="black"
)

plt.title("Daily Transport Cost Distribution")
plt.xlabel("Daily Transport Cost")
plt.ylabel("Number of Students")

plt.show()
```

##### Q5. Compare average internet usage by gender.

```python
internet_by_gender = students.groupby("gender")["internet_usage"].mean()

print(internet_by_gender)

internet_by_gender.plot(
    kind="bar",
    figsize=(8, 5),
    color=["pink", "skyblue"]
)

plt.title("Average Internet Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Internet Usage")

plt.xticks(rotation=0)

plt.show()
```

Male students have higher average internet usage in this dataset.
